```python
text = " Ｕｎｉｃｏｄｅ! 🅤🅝🅘🅒🅞🅓🅔‽ 🇺‌🇳‌🇮‌🇨‌🇴‌🇩‌🇪! 😄 The very name strikes fear and awe into the hearts of programmers worldwide. We all know we ought to “support Unicode” in our software (whatever that means—like using wchar_t for all the strings, right?). But Unicode can be abstruse, and diving into thethousand-page Unicode Standard plus its dozens of supplementary annexes, reports, and notes can be more than a little intimidating. I don’t blame programmers for still finding the whole thing mysterious, even 30 years after Unicode’s inception."
tokens = text.encode("utf-8")
tokens = list(map(int , tokens))
print(text)
print(len(text))
print("=================")
print(tokens)
print(len(tokens));
```

```python
def get_stats(ids):
  counts = {}
  for pair in zip(ids ,ids[1:]):
    counts[pair] = counts.get(pair ,0)+1
  return counts
stats = get_stats(tokens)
print(stats)
```

```python
print(sorted(((v,k) for k , v in stats.items()), reverse = True))
```

```python
chr(101) , chr(32)
```

```python
top_pair = max(stats , key=stats.get)
print(top_pair , stats[top_pair])
```

```python
def merge(ids , pair , new_id):
  new_ids = []
  i = 0
  while i < len(ids):
    if i < len(ids)-1 and ids[i]== pair[0] and ids[i+1] == pair[1] :
     new_ids.append(new_id)
     i +=2
    else:
      new_ids.append(ids[i])
      i+=1

  return new_ids

```

```python
tokens2 = merge(tokens , top_pair , 256)
tokens2.count(256)
```

```python
num_range = 100;
for i in range(num_range):
  stats = get_stats(tokens)
  if not stats:
    break

  pair = max(stats , key=stats.get)
  new_ids = 256+i
  tokens = merge(tokens , pair , new_ids)
```
